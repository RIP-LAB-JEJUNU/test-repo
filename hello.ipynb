## 3. 데이터 다뤄보기 (pandas)

Colab에는 `pandas`, `numpy`, `matplotlib`이 이미 설치되어 있어 바로 쓸 수 있습니다.

```python
import pandas as pd

df = pd.DataFrame(
    {
        "이름": ["가람", "나린", "다올", "라온", "마루"],
        "점수": [88, 95, 72, 64, 91],
    }
)

df["등급"] = pd.cut(
    df["점수"],
    bins=[0, 69, 79, 89, 100],
    labels=["D", "C", "B", "A"],
)

print("평균 점수:", df["점수"].mean())
df
```

## 4. 그래프 그리기 (matplotlib)

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2 * np.pi, 200)

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(x, np.sin(x), label="sin(x)")
ax.plot(x, np.cos(x), label="cos(x)", linestyle="--")
ax.set_title("sin & cos")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend()
ax.grid(alpha=0.3)

plt.show()
```
